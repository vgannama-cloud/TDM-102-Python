## Filtering for a specific Olympic team, event and year

```python
olympics_df.shape
```

This tells us how many athlete records (rows) and how many pieces of information (columns) we have in the Olympics dataset

```python
olympics_df.tail()
```

The tail shows the last 5 rows of the dataset

```python
jamaican_bobsled = olympics_df[(olympics_df['Team'] == 'Jamaica') & 
                                (olympics_df['Event'].str.contains('Bobsleigh')) & 
                                (olympics_df['Year'] == 1988) & 
                                (olympics_df['Season'] == 'Winter')]
jamaican_bobsled
```

Filter the data using multiple conditions with the & (AND) operator

This shows the 4 Jamaican athletes who competed in Bobsleigh at the 1988 Winter Olympics. There are 6 rows because 2 athletes competed in both the Two-Man and Four-Man Bobsleigh events. The athletes are: Dudley Stokes, Devon Harris, Michael White, and Nelson Stokes.

## Finding the top 5 countries by Olympic athlete entries

```python
team_counts = olympics_df['Team'].value_counts()
team_counts
```

This shows all countries ranked by total number of athlete entries. The United States, France, Great Britain, Italy, and Germany have the most.

```python
top5_teams = olympics_df['Team'].value_counts().head(5)
top5_teams
```

This displays only the top 5 results which are the countries with most athlete entries

```python
import matplotlib.pyplot as plt
plt.bar(top5_teams.index, top5_teams.values)
plt.title('Top 5 Countries by Number of Olympic Athlete Entries')
plt.xlabel('Country')
plt.ylabel('Number of Athlete Entries')
plt.show()
```

This bar plot visualizes which countries have the most athlete entries. The height of each bar shows the total number of times athletes from that country appear. This includes all Olympic games in the dataset, across all years and events.

## Loading restaurant vendor data and counting by category

```python
vendors_df = pd.read_csv("/anvil/projects/tdm/data/restaurant/vendors.csv")
vendors_df.head()
```

```python
vendor_counts = vendors_df['vendor_category_en'].value_counts()
vendor_counts
```

This shows the two types of vendors in the dataset and how many times each appears. ``value_counts()`` gives us the frequency of each unique value in the column.


## Visualizing vendor counts by category as a bar chart

```python
plt.bar(vendor_counts.index, vendor_counts.values)
plt.title('Number of Vendors by Category')
plt.xlabel('Vendor Category')
plt.ylabel('Number of Vendors')
plt.show()
```

This bar plot has only two bars, one for each vendor type. The height of each bar shows how many vendors belong to that category.
