## Average airtime for specific origin-destination pairs

```python
import pandas as pd
myDF = pd.read_csv('/anvil/projects/tdm/data/flights/subset/2006.csv')
```

```python
pd.set_option('display.max_columns', None)
```

To Ensure all columns are visible

```python
pa_flights = myDF[myDF['Dest'] == 'PHL']
```

This is the subset for flights with Philadelphia (PHL) as the destination

```python
num_phl_origins = pa_flights['Origin'].nunique()
print(f"Number of flight origins for PHL in 2006: {num_phl_origins}")
```

```python
phl_airtimes = pa_flights.groupby('Origin')['AirTime'].mean().sort_values(ascending=False)
sfo_avg_airtime = phl_airtimes['SFO']
print(f"Average AirTime from SFO to PHL: {sfo_avg_airtime}")
```

Find average AirTime for each origin going to PHL, sorted descending. We then locate 'SFO' (San Francisco) in that result

```python
az_flights = myDF[myDF['Dest'] == 'PHX']
phx_airtimes = az_flights.groupby('Origin')['AirTime'].mean().sort_values(ascending=False)
hnl_avg_airtime = phx_airtimes['HNL']
print(f"Average AirTime from HNL to PHX: {hnl_avg_airtime}")
```

This creates subset for Phoenix (PHX) and find average AirTime from HNL

## Bucketing flight departure and arrival times into parts of the day

```python
my_flights = pd.read_csv('/anvil/projects/tdm/data/flights/subset/1997.csv')
```

```python
time_bins = [0, 300, 600, 900, 1200, 1500, 1800, 2100, 2400]
time_labels = ["Late Night", "Early Morning", "Morning", "Late Morning", 
               "Afternoon", "Early Evening", "Evening", "Night"]
```

Define bins and labels for flight times

```python
my_flights['depart_times'] = pd.cut(my_flights['DepTime'], bins=time_bins, labels=time_labels)
my_flights['arrival_times'] = pd.cut(my_flights['ArrTime'], bins=time_bins, labels=time_labels)
```

Create depart_times column and arrival_times column

To identify busiest departure and arrival times

```python
print("Departure Time Counts:")
print(my_flights['depart_times'].value_counts(sort=False))
```

```python
print("\nArrival Time Counts:")
print(my_flights['arrival_times'].value_counts(sort=False))
```

```python
busiest_dep = my_flights['depart_times'].value_counts().idxmax()
busiest_arr = my_flights['arrival_times'].value_counts().idxmax()
```

```python
print(f"\nBusiest Departure Time: {busiest_dep}")
print(f"Busiest Arrival Time: {busiest_arr}")
```

## Total and average airtime by day of week and month

```python
march_flights = my_flights[my_flights['Month'] == 3]
print("Day of Week counts for March:")
print(march_flights['DayOfWeek'].value_counts().sort_index())
```

Compare DayOfWeek counts for March vs All Months

```python
print("\nDay of Week counts for All Months:")
print(my_flights['DayOfWeek'].value_counts().sort_index())
```

```python
total_airtime_table = my_flights.groupby(['DayOfWeek', 'Month'])['AirTime'].sum().unstack()
print("\nTotal AirTime Table:")
print(total_airtime_table)
```

Display total (sum) airtime across DayOfWeek and Month

```python
mean_airtime_table = my_flights.groupby(['DayOfWeek', 'Month'])['AirTime'].mean().unstack()
print("\nAverage AirTime Table:")
print(mean_airtime_table)
```

Display average (mean) airtime across DayOfWeek and Month
