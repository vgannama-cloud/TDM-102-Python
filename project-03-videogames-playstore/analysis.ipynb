## Average critic score by genre and total sales by console

```python
import pandas as pd
video_games = pd.read_csv("/anvil/projects/tdm/data/grouping/vgchartz-2024.csv")
video_games.head()
```

```python
video_games.shape
```

```python
genre_scores = video_games.groupby("genre", as_index=False)["critic_score"].mean()
genre_scores
```

This shows the average critic score for each genre. Python automatically ignores NA values when calculating the mean.

Here we can see that Sandbox has the highest average critic score.

```python
console_sales = video_games.groupby("console", as_index=False)["total_sales"].sum()
console_sales.sort_values("total_sales", ascending=False)
```

This shows the total sales (in millions) for each console.

As displayed in the table, PS2 has the highest number of total sales.

The ``.groupby()`` function in Python allows us to split our dataset into groups based on the values in one or more columns. After grouping, we can apply functions like ``.mean()`` or ``.sum()`` to calculate statistics for each group. This is useful because it lets us compare different categories efficiently without having to filter the data manually for each category.

## Flagging high and low rated games and finding standout sports titles

```python
video_games['high_rated'] = video_games['critic_score'] >= 7
```

This creates a True/False column. Rows with critic_score of 7 or more get True, others get False.

```python
video_games['low_rated'] = video_games['critic_score'] < 7
```

This shows the same for games with critic_score strictly less than 7.

```python
video_games['high_rated'].value_counts()
```

```python
video_games['low_rated'].value_counts()
```

```python
sports_hits = video_games[(video_games['genre'] == 'Sports') & 
                          (video_games['high_rated'] == True)]
sports_hits[sports_hits['critic_score'] == 9.5]
```

This shows all Sports games that received exactly 9.5.

```python
sports_fails = video_games[(video_games['genre'] == 'Sports') & 
                           (video_games['low_rated'] == True)]
sports_fails[sports_fails['critic_score'] == 2.8]
```

This shows all Sports games that received exactly 2.8.

## Cleaning malformed review counts in the Google Play Store data

```python
app_store = pd.read_csv("/anvil/projects/tdm/data/grouping/googleplaystore.csv")
app_store.head()
```

```python
app_store.dtypes
```

We can see that 'Reviews' is 'object' (string) instead of numeric.

```python
app_store["Clean_Reviews"] = app_store["Reviews"].str.replace(",", "", regex=False)
app_store["Numeric_Reviews"] = app_store["Clean_Reviews"].astype(float)
```

The error message at the bottom shows "3.0M" is causing the problem. This value represents 3.0 million but cannot be converted to float.

```python
app_store = app_store[app_store["Clean_Reviews"] != "3.0M"]
app_store["Numeric_Reviews"] = app_store["Clean_Reviews"].astype(float)
```

We filter out rows where Clean_Reviews equals "3.0M".

```python
app_store["rating_per_review"] = app_store["Rating"] / app_store["Numeric_Reviews"]
app_store["rating_per_review"].head()
```

This creates a new column showing rating divided by number of reviews.

```python
app_store["Content Rating"].unique()
app_store["Content Rating"].unique().shape
```

```python
app_store.groupby("New_Content_Ratings", as_index=False)["App"].count()
```
