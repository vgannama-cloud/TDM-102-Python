## Loading the dataset and plotting the distribution of movie ratings

```python
import pandas as pd
myDF = pd.read_csv("/anvil/projects/tdm/data/movies_and_tv/rotten_tomatoes_movies.csv")
myDF.head()
```

```python
myDF.tail()
```

```python
myDF.shape
```

There are 16638 rows and 23 columns in this dataset.


```python
pd.set_option('display.max_columns', None)
myDF.head()
```

Now we can see all 23 columns in the output. Before this setting, pandas might have hidden some middle columns with "...".

```python
my_rating_counts = myDF['rating'].value_counts()
my_rating_counts
```

This shows the distribution of movies across rating categories. R has the most movies (6136), followed by NR (Not Rated) with 4862. We can see there are some typos like 'PG-13)' and 'R)' that need cleaning.

```python
import matplotlib.pyplot as plt
plt.plot(my_rating_counts, 
         color='red',
         linewidth=2,
         marker='o')
plt.title('Distribution of Movies by Rating Category')
plt.xlabel('Rating Category')
plt.ylabel('Number of Movies')
plt.show()
```

This plot visualizes how movies are distributed across rating categories. The customizations include: red color, thicker line (width=2), and circular markers.

## Matching movies by genre, including partial genre matches

```python
comedy_exact = myDF[myDF['genre'] == "Comedy"]
comedy_exact.head()
```

The ``==`` operator finds exact matches only. These movies have "Comedy" listed as their sole genre.

```python
num_comedy_exact = comedy_exact.shape[0]
num_comedy_exact
```

Using ``.shape[0]`` gives us the number of rows. This is the count of movies that are ONLY labeled as "Comedy".

```python
comedy_contains = myDF[myDF['genre'].str.contains("Comedy", na=False)]
comedy_contains.head()
```

This includes movies like "Comedy, Drama" or "Comedy, Romance". The ``na=False`` parameter handles any missing values in the genre column.

```python
num_comedy_total = myDF['genre'].str.contains("Comedy", na=False).sum()
num_comedy_total
```

```python
comedy_not_exact = myDF[(myDF['genre'].str.contains("Comedy", na=False)) & 
                         (myDF['genre'] != "Comedy")]
comedy_not_exact.head()
```

These are movies with "Comedy" plus other genres.

```python
romance_contains = myDF[myDF['genre'].str.contains("Romance", na=False)]
romance_contains.head()
```

This shows romantic movies, including combinations like "Romance, Drama".

## Fixing inconsistent rating labels and cross tabulating rating vs critical status

```python
myDF['rating'].value_counts()
```

```python
myDF.loc[myDF['rating'] == 'R)', 'rating'] = 'R'
myDF.loc[myDF['rating'] == 'PG-13)', 'rating'] = 'PG-13'
```

``loc[condition, column]`` value replaces values that meet the condition. This fixes human typing errors

```python
myDF['rating'].value_counts()
```

```python
myDF['tomatometer_status'].value_counts()
```

This shows three categories: Rotten, Fresh, and Certified Fresh. Looking at the table, Rotten tomatometer status was given to most films

```python
table_data = pd.crosstab(myDF['rating'], myDF['tomatometer_status'])
table_data
```

This creates a table with rating as rows and tomatometer_status as columns
