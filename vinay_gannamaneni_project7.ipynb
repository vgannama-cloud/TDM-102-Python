## Classifying social media usage vs sleep with nested conditions

```python
import pandas as pd
import numpy as np

myDF = pd.read_csv('/anvil/projects/tdm/data/social_media_addiction/social_media.csv')
```

```python
max_sleep = myDF['Sleep_Hours_Per_Night'].max()
max_social = myDF['Avg_Daily_Usage_Hours'].max()

print(f"Longest recorded sleep time: {max_sleep} hours")
print(f"Longest social media time: {max_social} hours")
```

We can use ``.max()`` to find max values

```python
myDF['Status'] = np.where(myDF['Avg_Daily_Usage_Hours'] > myDF['Sleep_Hours_Per_Night'], "Doom Scroll",
                         np.where(myDF['Avg_Daily_Usage_Hours'] < myDF['Sleep_Hours_Per_Night'], "Healthy",
                                 "Balanced"))

print(myDF['Status'].value_counts())
```

Create the Status column using nested ``np.where()`` and check which habit is most common

## While loops counting down screen time and up sleep time

```python
screen_time = 10
while screen_time > 2:
    print(f"Screen time: {screen_time} hours")
    screen_time = screen_time - 1
print(f"Final screen time: {screen_time}")
```

```python
sleep_time = 2
while sleep_time < 10:
    print(f"Sleep time: {sleep_time} hours")
    sleep_time = sleep_time + 1
print(f"Final sleep time: {sleep_time}")
```

These are the iterative results from the screen_time loop, and the sleep_time loop

```python
screen_time = 10
sleep_time = 2

while screen_time > 2:
    print(f"Current values -> Screen: {screen_time}, Sleep: {sleep_time}")
    screen_time = screen_time - 0.5
    sleep_time = sleep_time + 0.5

print(f"Loop finished. Final Screen: {screen_time}, Final Sleep: {sleep_time}")
```

We then combined the loops with 0.5 increments

## Looping through Star Wars characters to count human vs non-human species

```python
characters = pd.read_csv('/anvil/projects/tdm/data/starwars/characters.csv')
```

```python
i = 0
char_count = 0

while char_count < 20:
    species_val = characters.loc[i, "species"]
    print(f"Row {i}: Species is {species_val}")
    
    i += 1
    char_count += 1
```

This shows us the species of the first 20 characters.

```python
i = 0
char_count = 0
human_count = 0
nonhuman_count = 0

while char_count < 20:
    species = characters.loc[i, "species"]
    if species == "Human":
        print(f"Character number {i} ({characters.loc[i, 'name']}) is a human")
        human_count += 1
    else:
        print(f"Character number {i} ({characters.loc[i, 'name']}) is NOT a human")
        nonhuman_count += 1
    
    i += 1
    char_count += 1
```

Made counters to determine which species are human and non human

```python
print(f"Number of Humans: {human_count}")
print(f"Number of Non-Humans: {nonhuman_count}")
```

## Using a while loop to sum mental health scores

- while [condition]: checks the condition before running the code block. If the condition is false at the start, the loop never runs.
- while True: creates an infinite loop that runs until a break statement is triggered inside the block. It is useful when you want to perform an action at least once or check the exit condition in the middle of the logic.

```python
i = 0
total_count = 0

print(f"Starting mental health score is {total_count}\n")

while True:
    # Access the Mental_Health_Score for the current student index i
    student_score = myDF["Mental_Health_Score"][i]
    print(f"Mental health of student {i} is {student_score}")
    
    # Update the running total
    total_count = total_count + student_score
    print(f"Current mental health score is {total_count}\n")

    # Exit condition
    if total_count >= 100:
        print("All done!!!!")
        break

    i = i + 1
```

This is the iterative output of counting up to the final mental health score.
