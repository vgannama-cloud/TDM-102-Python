## Recoding survival into categories and combining with sex

```python
myDF["Status"] = pd.Categorical(
    myDF["Survived"],
    categories=[0, 1],
    ordered=True
).rename_categories(["Dead", "Alive"])
```

This converts 0 to "Dead" and 1 to "Alive". The ``ordered=True`` preserves this as an ordered category.

```python
myDF['Status'].value_counts()
```

This shows us how many passengers were Dead vs Alive.

```python
myDF["Combined"] = myDF["Status"].astype(str).str.cat(myDF["Sex"], sep=", ")
```

``.str.cat()`` combines the two columns with ", " as separator.

```python
myDF['Combined'].value_counts()
```

I used ``pd.Categorical`` with 'rename_categories' to create the Status column. This method preserves the ordered nature of the categories.

## Exploring the age range and identifying older passengers

```python
old_ages = [61, 62, 63, 64, 65, 66, 70, 70.5, 71, 74, 80]
old = myDF["Age"][myDF["Age"].isin(old_ages)]
```

```python
my_list1 = myDF["Age"][myDF["Age"] <= 10]
my_list1
```

## Grouping passengers into age brackets and cross tabulating with survival status

```python
children = myDF["Age"][myDF["Age"] <= 20]
children.value_counts()
```

This contains all ages 20 and under and shows the distrubution of children's ages

```python
print("Children min:", children.min())
print("Children max:", children.max())
```

Confirms the range is correct

```python
young = myDF["Age"][(myDF["Age"] > 20) & (myDF["Age"] <= 40)]
young.value_counts()
```

This contains ages from 21 to 40.

```python
print("Young min:", young.min())
print("Young max:", young.max())
```

Confirms the range is 21-40.

```python
adult = myDF["Age"][(myDF["Age"] > 40) & (myDF["Age"] <= 60)]
adult.value_counts()
```

```python
print("Adult min:", adult.min())
print("Adult max:", adult.max())
```

Confirms the range is 41-60.
